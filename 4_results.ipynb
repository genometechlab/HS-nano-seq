# Results — confusion matrices for every run

One section per run listed in `experiments.sh`. Plotting lives in `4_results.py`
(`python 4_results.py` writes the same figures to `figures/` without Jupyter).

- **PH009/9fea1263** is the training run: `Data/model/split_dict.pkl` splits its reads
  into train / val / test, one panel each.
- **Every other run** (PH009/266e1025, PH015/cd8c3e45, PH015/78073ab4) is test-only:
  one panel over all its reads.

Matrices are row-normalized (diagonal = per-class recall) and annotated with recall
and raw count. Axes show the class as its PH009 barcode name (class 0 → BC01 …
class 15 → BC16); PH015 classes come from barcode09–24, as noted in each title.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import importlib
results = importlib.import_module("4_results")   # name starts with a digit: no plain import
load_config, load_split, load_preds, plot_run, print_summary = (
    results.load_config, results.load_split, results.load_preds, results.plot_run, results.print_summary)

cfg = load_config()                       # runs + paths from experiments.sh
split_rids = load_split(cfg["split_dict"])
runs = {r["key"]: r for r in cfg["runs"]}
rows = []                                 # summary rows, filled by show()


def show(key):
    preds = load_preds(cfg["processed"], key)
    if preds is None:
        return
    fig, r = plot_run(runs[key], preds, split_rids, cfg["train_run"])
    rows.extend(r)
    plt.show()
```
